```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024] = 0 M 8

sales = np.array([[1, 150000, 180000, 220000, 250000],# Paradise Biryani
[2, 120000, 140000, 160000, 190000], # Beijing Bites
[3, 200000, 230000, 260000, 300000], # Pizza Hub
[4, 180000, 210000, 240000, 270000], # Burger Point
[5, 160000, 185000, 205000, 230000]]) # Chai Point


print(" ==== Zomato sales analysis ==== ")
print("\n Sales data shape", sales. shape)
print("\n Sample data for 1st 3 : \n", sales[0:3])
print("\nTotal Sales Per Year: ",sum(sales[:,1:]))
```

```python
print("Max Sales per Year: ",np.max(sales[:,1:],axis=1))
print("Max Sales per Item: ",np.max(sales[:,1:],axis=0))

print("Min Sales per Year: ",np.min(sales[:,1:],axis=1))
print("Min Sales per Item: ",np.min(sales[:,1:],axis=0))

print("Mean Sales per Year: ",np.mean(sales[:,1:],axis=1))
print("Mean Sales per Item: ",np.mean(sales[:,1:],axis=0))
```

```python
cumulative_sales = np.cumsum(sales[:,1:],axis=1)
print(cumulative_sales)

plt.plot(np.mean(cumulative_sales,axis=0))
plt.title("Average Cumulative Sales by Year")
plt.xlabel("Years")
plt.ylabel("Average Cumulative Sales")
plt.grid(True)
plt.show()
```
