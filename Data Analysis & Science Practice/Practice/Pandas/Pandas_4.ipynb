```python
import numpy as np
import pandas as pd
```

```python
employees = pd.DataFrame({
    "EmpID": [101, 102, 103, 104],
    "Name": ["John", "Alice", "Bob", "David"],
    "DeptID": [1, 2, 1, 3]
})
```

```python
departments = pd.DataFrame({
    "DeptID": [1, 2, 3,4],
    "Department": ["HR", "IT", "Finance","Testing"]
})
```

```python
employees
```

```python
departments
```

```python
pd.merge(employees,departments)
```

```python
pd.merge(employees,departments,on='DeptID')
```

```python
pd.merge(employees,departments,on='DeptID',how='inner')
```

```python
pd.merge(employees,departments,on='DeptID',how='outer')
```

```python
pd.concat((employees,departments))
```

```python
pd.concat((employees,departments),axis=1)
```

```python
employees.set_index('DeptID').join(departments.set_index('DeptID'))
```
