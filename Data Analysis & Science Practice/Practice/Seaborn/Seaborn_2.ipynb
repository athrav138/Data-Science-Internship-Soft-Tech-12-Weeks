```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
penguins = sns.load_dataset('penguins')
```

```python
penguins
```

```python
penguins.dropna()
```

```python
# Scatter Plot and Differnt Attributes
sns.scatterplot(data=penguins,x='flipper_length_mm',y='body_mass_g',hue='sex')
sns.set_style('darkgrid')
sns.despine(left=False)
sns.set_theme("paper")
sns.set_context("poster")
sns.set_palette('muted')

plt.show()
```

```python
# Bar Plot
sns.barplot(data=penguins,x='flipper_length_mm',y='body_mass_g',hue='sex')
sns.set_style('darkgrid')
sns.despine(left=False)
sns.set_theme("paper")
sns.set_context("poster")
sns.set_palette('deep')

plt.show()
```

```python
# scatter Plot Distribution overlapping
sns.scatterplot(data=penguins,x='species',y='body_mass_g',hue='island')
sns.set_context('paper')
sns.set_theme('paper')

```

```python
# Swarm plot for solution  of previous scatter plot
sns.swarmplot(data=penguins,x='species',y='body_mass_g',hue='island')
```

```python
sns.histplot(data=penguins,x="body_mass_g",hue='sex',multiple="stack")
```

```python
# Regression Plot
sns.regplot(data=penguins,x='flipper_length_mm',y='body_mass_g')
plt.show()
```

```python
sns.lineplot(data=penguins,x='flipper_length_mm',y='body_mass_g',hue='island',style='sex')
```

```python
sns.jointplot(data=penguins,x='flipper_length_mm',y='body_mass_g')
```

```python
sns.jointplot(data=penguins,x='flipper_length_mm',y='body_mass_g',hue='sex')
```

```python
sns.jointplot(data=penguins,x='flipper_length_mm',y='body_mass_g',kind='reg')
```

```python
sns.barplot(data=penguins,x='species',y='body_mass_g',hue='sex')
```

```python
sns.boxplot(data=penguins,x='species',y='body_mass_g',hue='sex')
```

```python
sns.violinplot(data=penguins,x='species',y='body_mass_g',hue='sex',split=True)
```

```python
# combination of the violen and swarm plot

sns.violinplot(data=penguins,x='species',y='body_mass_g')
sns.swarmplot(data=penguins,x='species',y='body_mass_g',hue='island')
```

```python
# kde (Kernel Density Plot)

sns.kdeplot(data=penguins,x='body_mass_g',hue='species',fill='True')
```
