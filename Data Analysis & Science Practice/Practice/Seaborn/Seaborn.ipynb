```python
import matplotlib.pyplot as plt
import seaborn as sns

x = [12,23,45,57,87,34,23,56,45]

sns.histplot(x,bins=3,kde=True)
plt.title("Histogram")
plt.show()
```

```python
sns.boxplot(data=x)
plt.title("Boxplot")
plt.show()
```

```python
data = pd.DataFrame({
    "marks":[12,443,46,42,3],
    "percent":[34,65,223,423,34],
    "heat":[34,67,34,67,56]
    })

sns.heatmap(data)
plt.title("Heatmap")
plt.show()
```
