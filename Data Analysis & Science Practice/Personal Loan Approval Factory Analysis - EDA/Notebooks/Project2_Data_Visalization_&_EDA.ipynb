```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("cleaned1.csv")
```

```python
df
```

### Data Visualization using Matplotlib

```python
# 1. Histogram for Age Distribution
plt.figure(figsize=(10, 6))
plt.hist(df['age'], bins=20, color='skyblue', edgecolor='black')
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.tight_layout()
plt.savefig('age_distribution.png')
plt.show()

# Insight: The age distribution shows a varied range, with perhaps a slightly higher frequency in certain age groups.
```

```python
# 2. Histogram for Annual Income Distribution
plt.figure(figsize=(10, 6))
plt.hist(df['annual_income'], bins=30, color='lightcoral', edgecolor='black')
plt.title('Distribution of Annual Income')
plt.xlabel('Annual Income')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.tight_layout()
plt.savefig('annual_income_distribution.png')
plt.show()

# Insight: The annual income distribution is skewed to the right, indicating most individuals have lower to mid-range incomes, with fewer having very high incomes.
```

```python
# 3. Histogram for Credit Score Distribution
plt.figure(figsize=(10, 6))
plt.hist(df['credit_score'], bins=20, color='lightgreen', edgecolor='black')
plt.title('Distribution of Credit Score')
plt.xlabel('Credit Score')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.tight_layout()
plt.savefig('credit_score_distribution.png')
plt.show()

# Insight: Credit scores appear to be somewhat normally distributed, centered around a certain range, which is typical for such datasets.
```

```python
# 4. Bar Plot for Occupation Status Distribution
occupation_counts = df['occupation_status'].value_counts()
plt.figure(figsize=(12, 7))
plt.bar(occupation_counts.index, occupation_counts.values, color='gold', edgecolor='black')
plt.title('Distribution of Occupation Status')
plt.xlabel('Occupation Status')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.75)
plt.tight_layout()
plt.savefig('occupation_status_distribution.png')
plt.show()

# Insight: This plot shows the most common occupation statuses among the customers, highlighting the primary employment categories.
```

```python
# 5. Bar Plot for Loan Status (Target Variable)
loan_status_counts = df['loan_status'].value_counts()
plt.figure(figsize=(8, 6))
plt.bar(loan_status_counts.index.astype(str), loan_status_counts.values, color=['green', 'red'], edgecolor='black')
plt.title('Distribution of Loan Status (0: Approved, 1: Declined)')
plt.xlabel('Loan Status')
plt.ylabel('Count')
plt.xticks([0, 1], ['Approved', 'Declined'])
plt.grid(axis='y', alpha=0.75)
plt.tight_layout()
plt.savefig('loan_status_distribution.png')
plt.show()

# Insight: The distribution of loan status reveals the balance between approved and declined loans, which is crucial for understanding the target variable.
```

```python
# 6. Scatter Plot: Annual Income vs. Credit Score
plt.figure(figsize=(12, 8))
plt.scatter(df['annual_income'], df['credit_score'], alpha=0.5, color='purple')
plt.title('Annual Income vs. Credit Score')
plt.xlabel('Annual Income')
plt.ylabel('Credit Score')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.savefig('income_vs_credit_score.png')
plt.show()

# Insight: This scatter plot helps visualize if there's a correlation between a person's income and their credit score. Higher income might generally correlate with higher credit scores, but exceptions exist.
```

### Conclusion from Visualizations:

*   **Age and Income:** The age distribution is spread out, while annual income is right-skewed, suggesting a common pattern where most individuals fall into lower-to-middle income brackets.
*   **Credit Score:** Credit scores show a relatively normal distribution, which indicates a typical range of creditworthiness within the dataset.
*   **Occupation Status:** The bar chart for `occupation_status` highlights the predominant employment categories, which can be useful for segmenting customers.
*   **Loan Status:** The `loan_status` distribution provides an initial understanding of the balance between approved and declined loans. Further analysis is needed to identify factors influencing this outcome.
*   **Income vs. Credit Score:** The scatter plot for `annual_income` vs `credit_score` can indicate trends or clusters. A positive correlation would suggest that higher incomes are generally associated with better credit scores, but it's important to look for outliers or groups that deviate from this trend.
