```python
import pandas as pd
import numpy as np
```

```python
arr = np.array([10,20,30])
list = [10,20,30]
labels = ["a","b","c"]
dict = {1:10,2:20,3:30}

```

```python
s1 = pd.Series(arr)
print(s1)
s1 = pd.Series(arr,index=labels)
print(s1)
```

```python
s2 = pd.Series(list)
print(s2)
s2 = pd.Series(list,index=labels)
print(s2)
```

```python
s3 = pd.Series(dict)
print(s3)
```
