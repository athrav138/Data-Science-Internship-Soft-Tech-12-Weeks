```python
import matplotlib.pyplot as plt
days_of_week = ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
sales = [34,55,61,34,40,20,70]

plt.plot(days_of_week,sales,color='red',label='trend',linestyle='--',marker='*',linewidth=2)
plt.title('Sales of Icecream week')
plt.xlabel('Days')
plt.ylabel('Sales per Day')
plt.grid()
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
x = ['A','B','C','D']
y = [1200,1000,1400,700]


plt.bar(x,y,color='Orange',label='Sales 2024',)
plt.xlabel('Product')
plt.title('Sales in 2025')
plt.ylabel('Sales')
plt.legend()
plt.show()
```

```python
# pie chart

regions = ['North','East','West','South']
value = [12000,30000,40000,40000]
plt.pie(value,colors=['red','blue','orange','green'],labels=regions,autopct='%1.1f%%')
plt.title('Region Sales')
plt.legend()

plt.show()
```

```python
scores = [98,56,42,56,87,94,83,82,37,56,45,84,76,34]
plt.hist(scores,bins=4,color='red',edgecolor='black')
plt.xlabel('Category')
plt.ylabel('Values')
plt.title('Sales')
plt.show()
```

```python
x = [1,2,3,4,5]
y = [10,20,30,40,50]
z = [14,24,34,44,54]

plt.scatter(x,y,color='red',label='2025')
plt.scatter(x,z,color='blue',label='2026')

plt.xlabel('Category')
plt.ylabel('values')
plt.title('Sales')
plt.grid()
plt.legend()
plt.show()

```
