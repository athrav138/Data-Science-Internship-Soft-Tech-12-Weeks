```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [10,20,30,40,50]

plt.subplot(1,2,1)
plt.plot(x,y)
plt.title("Line Chart")
plt.tight_layout()

plt.subplot(1,2,2)
plt.bar(x,y)
plt.title("bar chart")

plt.subplot(2,2,3)
plt.barh(x,y)
plt.title("bar horizontal chart")

plt.subplot(2,2,4)
plt.scatter(x,y)
plt.title("scatter chart")
plt.show()
```

```python
fig,ax = plt.subplots(2,2,figsize=(10,5))
ax[0][0].plot(x,y)
ax[0][0].set_title("line chart")

ax[0][1].bar(x,y)
ax[0][1].set_title("bar chart")

ax[1][0].barh(x,y)
ax[1][0].set_title("bar horizontal chart")

ax[1][1].scatter(x,y)
ax[1][1].set_title("Scatter chart")

plt.suptitle("Charts in one Frame for comaparision")
plt.tight_layout()
plt.savefig("chart.png",dpi=300)

```
